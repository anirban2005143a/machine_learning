# Load document

```python
loader = UnstructuredPDFLoader(
    file_path=file_path,
    mode='elements',
    strategy = "hi_res"
)

docs_generator = loader.lazy_load()

docs = []

for doc in docs_generator : 
    print(doc)
    docs.append(doc)    
```

```python
print("Category:", docs[0].metadata.get("category"))
print("Page:", docs[0].metadata.get("page_number"))
print(docs[0].page_content[:300])
```

```python
categories = set()

for doc in docs : 
    if "category" in doc.metadata:
        categories.add(doc.metadata["category"])

print(categories)
```

```python
tables = [doc for doc in docs if doc.metadata.get("category") == "Table"]

for i, table in enumerate(tables):
    print(f"\n===== TABLE {i} =====")
    print(table.page_content)
```

# Print and add title to the metadata as heading

```python
from collections import defaultdict

sections = defaultdict(list)
current_heading = "NO HEADING"

for doc in docs:
    category = doc.metadata.get("category")

    if category in ["Title", "Heading"]:
        current_heading = doc.page_content.strip()
    elif category in ['ListItem', 'FigureCaption', 'NarrativeText']:
        sections[current_heading].append(doc.page_content)

    doc.metadata["heading"] = current_heading
```

```python
for doc in docs : 
    category = doc.metadata.get("category")

    if category == 'UncategorizedText':
        print(doc.page_content)
```

```python
for heading, texts in sections.items():
    print("\n" + "="*80)
    print("HEADING:", heading)
    print("-"*80)
    print("\n".join(texts[:3]))  # first 3 paragraphs
```

```python
sections = []
current_heading = "NO HEADING"

for doc in docs:
    cat = doc.metadata.get("category")
    if cat in ["Title", "Heading"]:
        current_heading = doc.page_content.strip()
    elif cat in ["NarrativeText", "ListItem", "Image", "Table"]:
        sections.append({
            "heading": current_heading,
            "category": cat,
            "content": doc.page_content,
            "element": getattr(doc, "element", None)  # images stored here
        })
```

# remove the title = "Image" document from docs

```python
filtered_doc = []

for doc in docs :
    if doc.metadata.get("category") not in ['Image' , 'UncategorizedText'] :
        filtered_doc.append(doc)
```

# text splitting

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=15,
    separators=[
        "\n\n",
        "\n",
        " ",
        ".",
        ",",
    ]
)

splitted_text = splitter.split_documents(filtered_doc)
```
