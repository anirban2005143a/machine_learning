```python
from langchain_huggingface import HuggingFaceEmbeddings , HuggingFaceEndpoint
from langchain_community.vectorstores import Chroma , FAISS
from langchain_core.documents import Document
from langchain_huggingface.embeddings import HuggingFaceEndpointEmbeddings
import os
from dotenv import load_dotenv
```

```python
doc1 = Document(
    page_content="""
Supervised learning is one of the most widely used paradigms in machine learning, where the model is trained on labeled data.
The algorithm learns to map input features to output labels by minimizing a loss function that quantifies the difference between
predicted and true values. Common supervised learning algorithms include linear regression, logistic regression, support vector
machines, decision trees, and neural networks. Applications span across image classification, spam detection, fraud detection,
and speech recognition. Performance evaluation often relies on metrics such as accuracy, precision, recall, F1-score, and ROC-AUC
to ensure the model generalizes well to unseen data.
""",
    metadata={"topic": "Supervised Learning"}
)

# 2️⃣ Unsupervised Learning
doc2 = Document(
    page_content="""
Unsupervised learning deals with datasets that do not contain labeled outputs, focusing on discovering hidden patterns or
intrinsic structures. Clustering algorithms, such as k-means, hierarchical clustering, and DBSCAN, group similar data points
based on feature similarity, while dimensionality reduction techniques like PCA, t-SNE, and UMAP reduce high-dimensional data
to more manageable representations. Unsupervised methods are widely used for customer segmentation, anomaly detection,
topic modeling, and data visualization. These algorithms help in understanding the data distribution and can often serve as
preprocessing steps for downstream supervised tasks.
""",
    metadata={"topic": "Unsupervised Learning"}
)

# 3️⃣ Deep Learning
doc3 = Document(
    page_content="""
Deep learning is a subset of machine learning that leverages artificial neural networks with multiple layers to model
complex patterns in large datasets. Convolutional neural networks (CNNs) excel at processing images, detecting spatial
hierarchies of features, while recurrent neural networks (RNNs) and their variants like LSTM and GRU are designed for
sequential data such as time series or text. Deep learning models automatically extract features from raw inputs, reducing
the need for manual feature engineering. Applications include image recognition, natural language processing, autonomous
driving, and medical diagnosis. Training deep networks requires substantial computational resources and large labeled datasets.
""",
    metadata={"topic": "Deep Learning"}
)

# 4️⃣ Transformers and LLMs
doc4 = Document(
    page_content="""
Transformers have revolutionized natural language processing and large language models (LLMs) by enabling efficient
attention mechanisms that process sequences in parallel. The transformer architecture consists of encoder and decoder blocks,
self-attention layers, and feed-forward networks, allowing models like BERT, GPT, and T5 to capture contextual relationships
in text. Pretrained LLMs leverage massive corpora and transfer learning to perform downstream tasks such as text
classification, summarization, question-answering, and text generation. Fine-tuning and prompt engineering further enhance
performance for specific applications, demonstrating the versatility and state-of-the-art capabilities of transformer-based models.
""",
    metadata={"topic": "Transformers and LLMs"}
)

# 5️⃣ Reinforcement Learning
doc5 = Document(
    page_content="""
Reinforcement learning (RL) is a machine learning paradigm where agents learn to make sequential decisions by interacting
with an environment to maximize cumulative rewards. The agent observes states, takes actions, and receives feedback in the
form of rewards or penalties. Key algorithms include Q-learning, SARSA, Deep Q-Networks (DQN), and policy gradient methods.
RL has been successfully applied to robotics, game playing, autonomous driving, and resource management. Recent advances
combine deep learning with RL, known as deep reinforcement learning, enabling agents to handle high-dimensional inputs such
as raw images and complex action spaces, pushing the boundaries of AI in decision-making tasks.
""",
    metadata={"topic": "Reinforcement Learning"}
)

# Combine all documents
docs = [doc1, doc2, doc3, doc4, doc5]
```

```python
load_dotenv()

from google.colab import userdata

embedding = HuggingFaceEndpointEmbeddings(
    model='sentence-transformers/all-MiniLM-L6-v2',
    huggingfacehub_api_token= userdata.get('HF_TOKEN')
)
```
