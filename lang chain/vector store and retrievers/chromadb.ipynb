```python
from langchain_community.vectorstores import Chroma
from langchain_huggingface import HuggingFaceEndpointEmbeddings
from langchain_core.documents import Document
from dotenv import load_dotenv
import os
```

```python
documents = [
    Document(
        page_content="Machine learning enables computers to learn patterns from data and make predictions.",
        metadata={"topic": "machine_learning"}
    ),
    Document(
        page_content="Cybersecurity focuses on protecting systems, networks, and data from digital attacks.",
        metadata={"topic": "cybersecurity"}
    ),
    Document(
        page_content="Climate change refers to long-term shifts in temperatures and weather patterns on Earth.",
        metadata={"topic": "climate_change"}
    ),
    Document(
        page_content="Blockchain is a distributed ledger technology that ensures transparent and secure transactions.",
        metadata={"topic": "blockchain"}
    ),
]
```

```python
embedding_model = HuggingFaceEndpointEmbeddings(
    model = 'sentence-transformers/all-MiniLM-L6-v2',
    huggingfacehub_api_token=os.environ['HF_TOKEN']
)
```

```python
vector_store = Chroma(
    embedding_function=embedding_model,
    persist_directory="chroma_db",
    collection_name="collection_1"
)
```

```python
ids = []

for doc in documents :
    ids.append(f"{len(ids)+1}_{doc.metadata['topic']}")

ids
```

```python
vector_store.add_documents(documents=documents , ids=ids)
```

```python
vector_store.similarity_search(
    query="probabilty and statictics",
    k=2
)
```

```python
vector_store.similarity_search(
    query="",
    filter={"topic" : "machine_learning"}
)
```
