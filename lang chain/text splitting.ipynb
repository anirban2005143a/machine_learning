```python
from langchain_text_splitters import CharacterTextSplitter , RecursiveCharacterTextSplitter
from langchain_community.document_loaders import PyPDFLoader
```

```python
loader = PyPDFLoader(file_path=r"D:\OneDrive - Indian Institute of Technology Indian School of Mines Dhanbad\Documents\books\System-Design-Alex-Xu-Vol-1.pdf")

docs_gen = loader.lazy_load()

docs = []

for document in docs_gen : 
    docs.append(document)

print(len(docs))
```

```python
import re

text = docs[50] + docs[51] + docs[52]

text = text.replace('\n' , '')
text = re.sub(r'\s+' , ' ' , text)
text
```

```python
spliter = CharacterTextSplitter(
    chunk_size = 100,
    chunk_overlap = 0,
    separator=''
)

# result = spliter.split_text(text = text)
result = spliter.split_documents(docs)
```
