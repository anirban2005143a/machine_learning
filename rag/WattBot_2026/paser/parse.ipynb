```python
from pathlib import Path

from mineru.parser import parse


PDF_PATH = Path(r"D:\machine learning\datasets\WattBot2026\data\fritze2021.pdf")
OUTPUT_DIR = Path("output")

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print(f"Parsing: {PDF_PATH}")

result = parse(
    PDF_PATH,
    tier="standard",
    ocr_mode="auto",
    page_range="all",
)

print("Parsing completed.")


# Markdown
markdown = result.markdown()

(OUTPUT_DIR / "paper.md").write_text(
    markdown,
    encoding="utf-8"
)


# Structured JSON
json_data = result.to_json()

(OUTPUT_DIR / "paper.json").write_text(
    json_data,
    encoding="utf-8"
)


print("Output:")
print(f"  {OUTPUT_DIR / 'paper.md'}")
print(f"  {OUTPUT_DIR / 'paper.json'}")
```
