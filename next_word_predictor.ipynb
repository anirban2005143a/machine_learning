```python
doc = doc.replace(',' , '')
doc = doc.replace('—' , '')
doc = doc.replace('“' , '')
doc = doc.replace('”' , '')
doc = doc.replace("'" , '')
```

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tokenizer = Tokenizer()
```

```python
tokenizer.fit_on_texts([doc])
```

```python
tokenizer.word_index
```

```python
len(tokenizer.word_index)
```

```python
tokenized_doc = []

for line in doc.split('.'):
  tokenized_doc.append(tokenizer.texts_to_sequences([line])[0])
```

```python
tokenized_doc[0:5]
```

```python
df_arr = []

for line in tokenized_doc :
  i = 0
  sen = []
  for i in range(0 , len(line)-1):
    sen.append(line[i])
    arr = [x for x in sen]
    arr.append(line[i+1])
    df_arr.append(arr)
    i= i +1
```

```python
df_arr[0:5]
```

```python
max_len = max([len(x) for x in df_arr])
max_len
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
paddeed_tokenized_doc = pad_sequences(df_arr , maxlen = max_len , padding = 'pre')
```

```python
paddeed_tokenized_doc
```

```python
x = paddeed_tokenized_doc[:,:-1]
y = paddeed_tokenized_doc[:,-1]
```

```python
x.shape
```

```python
from tensorflow.keras.utils import to_categorical
y = to_categorical(y , num_classes = len(tokenizer.word_index)+1)
```

```python
len(tokenizer.word_index)+1
```

```python
y.shape
```

```python
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense , LSTM , Input , Flatten , Embedding , BatchNormalization
```

```python
model = Sequential()

model.add(Input(shape=(x.shape[1] , )))

model.add(Embedding(len(tokenizer.word_index)+1 , 100 ))

model.add(LSTM(150 , return_sequences=True))
# model.add(LSTM(150 , return_sequences=True))
# model.add(LSTM(150 , return_sequences=True))
model.add(LSTM(150))

model.add(Dense(100 , activation='relu'))
model.add(BatchNormalization())

model.add(Dense(100 , activation='relu'))
model.add(BatchNormalization())

model.add(Dense(100 , activation='relu'))
model.add(BatchNormalization())

model.add(Dense(len(tokenizer.word_index)+1 , activation='softmax'))

model.summary()
```

```python
model.compile(loss='categorical_crossentropy' , optimizer = 'adam' , metrics = ['accuracy'])
```

```python
model.fit(x , y , epochs = 70)
```
