## Ensemble


```python
# ============================================================
# S6E10 - Feature Engineering + CatBoost + LightGBM + XGBoost
# + 5-Fold OOF Ensemble
# ============================================================

import warnings

warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd

from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score

from catboost import CatBoostClassifier
from lightgbm import LGBMClassifier
from xgboost import XGBClassifier

# ============================================================
# 1. LOAD DATA
# ============================================================

# Change these paths if required
train = pd.read_csv("../../datasets/kaggle-airline_passenger_satisfaction/train.csv")
test = pd.read_csv("../../datasets/kaggle-airline_passenger_satisfaction/test.csv")

TARGET = "satisfaction"

y = train[TARGET].copy()

X = train.drop(columns=[TARGET]).copy()
X_test = test.copy()


# ============================================================
# 2. REMOVE ID
# ============================================================

# ID normally should not be used as a predictive feature
if "id" in X.columns:
    X = X.drop(columns=["id"])

if "id" in X_test.columns:
    X_test = X_test.drop(columns=["id"])


# ============================================================
# 3. FEATURE ENGINEERING FUNCTION
# ============================================================


def feature_engineering(df):

    df = df.copy()

    # --------------------------------------------------------
    # Service columns
    # --------------------------------------------------------

    service_cols = [
        "Inflight wifi service",
        "Departure/Arrival time convenient",
        "Ease of Online booking",
        "Gate location",
        "Food and drink",
        "Online boarding",
        "Seat comfort",
        "Inflight entertainment",
        "On-board service",
        "Leg room service",
        "Baggage handling",
        "Check-in service",
        "Inflight service",
        "Cleanliness",
    ]

    existing_service_cols = [c for c in service_cols if c in df.columns]

    # --------------------------------------------------------
    # Overall service statistics
    # --------------------------------------------------------

    df["service_mean"] = df[existing_service_cols].mean(axis=1)
    df["service_sum"] = df[existing_service_cols].sum(axis=1)
    df["service_std"] = df[existing_service_cols].std(axis=1)
    df["service_min"] = df[existing_service_cols].min(axis=1)
    df["service_max"] = df[existing_service_cols].max(axis=1)

    df["service_range"] = df["service_max"] - df["service_min"]

    # --------------------------------------------------------
    # Service counts
    # --------------------------------------------------------

    df["good_service_count"] = (df[existing_service_cols] >= 4).sum(axis=1)

    df["bad_service_count"] = (df[existing_service_cols] <= 2).sum(axis=1)

    df["excellent_service_count"] = (df[existing_service_cols] == 5).sum(axis=1)

    df["poor_service_count"] = (df[existing_service_cols] <= 1).sum(axis=1)

    df["neutral_service_count"] = (df[existing_service_cols] == 3).sum(axis=1)

    # --------------------------------------------------------
    # Delay features
    # --------------------------------------------------------

    if (
        "Departure Delay in Minutes" in df.columns
        and "Arrival Delay in Minutes" in df.columns
    ):

        departure_delay = df["Departure Delay in Minutes"].fillna(0)

        arrival_delay = df["Arrival Delay in Minutes"].fillna(0)

        df["total_delay"] = departure_delay + arrival_delay

        df["delay_difference"] = arrival_delay - departure_delay

        df["delay_ratio"] = arrival_delay / (departure_delay + 1)

        df["has_departure_delay"] = (departure_delay > 0).astype(int)

        df["has_arrival_delay"] = (arrival_delay > 0).astype(int)

        df["has_any_delay"] = (df["total_delay"] > 0).astype(int)

        df["severe_departure_delay"] = (departure_delay >= 60).astype(int)

        df["severe_arrival_delay"] = (arrival_delay >= 60).astype(int)

        df["log_departure_delay"] = np.log1p(departure_delay)

        df["log_arrival_delay"] = np.log1p(arrival_delay)

        df["log_total_delay"] = np.log1p(df["total_delay"])

    # --------------------------------------------------------
    # Age features
    # --------------------------------------------------------

    if "Age" in df.columns:

        df["age_squared"] = df["Age"] ** 2

        df["age_young"] = (df["Age"] < 25).astype(int)

        df["age_middle"] = ((df["Age"] >= 25) & (df["Age"] < 50)).astype(int)

        df["age_senior"] = (df["Age"] >= 60).astype(int)

        df["age_group"] = pd.cut(
            df["Age"],
            bins=[0, 18, 25, 35, 45, 55, 65, 100],
            labels=False,
            include_lowest=True,
        )

    # --------------------------------------------------------
    # Flight distance
    # --------------------------------------------------------

    if "Flight Distance" in df.columns:

        df["log_flight_distance"] = np.log1p(df["Flight Distance"])

        if "Age" in df.columns:
            df["distance_per_age"] = df["Flight Distance"] / (df["Age"] + 1)

        df["short_flight"] = (df["Flight Distance"] < 500).astype(int)

        df["medium_flight"] = (
            (df["Flight Distance"] >= 500) & (df["Flight Distance"] < 2000)
        ).astype(int)

        df["long_flight"] = (df["Flight Distance"] >= 3000).astype(int)

    # --------------------------------------------------------
    # Categorical interactions
    # --------------------------------------------------------

    if "Type of Travel" in df.columns and "Class" in df.columns:
        df["travel_class"] = (
            df["Type of Travel"].astype(str) + "_" + df["Class"].astype(str)
        )

    if "Customer Type" in df.columns and "Class" in df.columns:
        df["customer_class"] = (
            df["Customer Type"].astype(str) + "_" + df["Class"].astype(str)
        )

    if "Customer Type" in df.columns and "Type of Travel" in df.columns:
        df["customer_travel"] = (
            df["Customer Type"].astype(str) + "_" + df["Type of Travel"].astype(str)
        )

    # --------------------------------------------------------
    # Service interactions
    # --------------------------------------------------------

    if "Inflight wifi service" in df.columns and "Online boarding" in df.columns:
        df["wifi_boarding"] = df["Inflight wifi service"] * df["Online boarding"]

    if "Seat comfort" in df.columns and "Inflight entertainment" in df.columns:
        df["comfort_entertainment"] = df["Seat comfort"] * df["Inflight entertainment"]

    if "Inflight service" in df.columns and "Cleanliness" in df.columns:
        df["service_cleanliness"] = df["Inflight service"] * df["Cleanliness"]

    if "Ease of Online booking" in df.columns and "Inflight wifi service" in df.columns:
        df["booking_wifi"] = df["Ease of Online booking"] * df["Inflight wifi service"]

    if "Food and drink" in df.columns and "Seat comfort" in df.columns:
        df["food_comfort"] = df["Food and drink"] * df["Seat comfort"]

    # --------------------------------------------------------
    # Higher-level experience features
    # --------------------------------------------------------

    digital_cols = [
        c
        for c in ["Inflight wifi service", "Ease of Online booking", "Online boarding"]
        if c in df.columns
    ]

    if digital_cols:
        df["digital_experience"] = df[digital_cols].mean(axis=1)

    comfort_cols = [
        c
        for c in ["Seat comfort", "Leg room service", "Cleanliness"]
        if c in df.columns
    ]

    if comfort_cols:
        df["comfort_score"] = df[comfort_cols].mean(axis=1)

    inflight_cols = [
        c
        for c in [
            "Food and drink",
            "Inflight entertainment",
            "Inflight service",
            "On-board service",
        ]
        if c in df.columns
    ]

    if inflight_cols:
        df["inflight_experience"] = df[inflight_cols].mean(axis=1)

    airport_cols = [
        c
        for c in [
            "Departure/Arrival time convenient",
            "Gate location",
            "Check-in service",
        ]
        if c in df.columns
    ]

    if airport_cols:
        df["airport_experience"] = df[airport_cols].mean(axis=1)

    return df


# ============================================================
# 4. APPLY FEATURE ENGINEERING
# ============================================================

X = feature_engineering(X)
X_test = feature_engineering(X_test)


# Make sure train/test have identical columns
X_test = X_test.reindex(columns=X.columns, fill_value=np.nan)


# ============================================================
# 5. CLEAN NUMERICAL VALUES
# ============================================================

numeric_cols = X.select_dtypes(include=["number"]).columns

X[numeric_cols] = X[numeric_cols].replace([np.inf, -np.inf], np.nan)

X_test[numeric_cols] = X_test[numeric_cols].replace([np.inf, -np.inf], np.nan)

for col in numeric_cols:

    median_value = X[col].median()

    X[col] = X[col].fillna(median_value)
    X_test[col] = X_test[col].fillna(median_value)


# ============================================================
# 6. CATEGORICAL COLUMNS
# ============================================================

categorical_cols = X.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

for col in categorical_cols:

    X[col] = X[col].astype(str)
    X_test[col] = X_test[col].astype(str)


# ============================================================
# 7. CV
# ============================================================

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)


# ============================================================
# 8. CATBOOST
# ============================================================

oof_cat = np.zeros(len(X))
test_cat = np.zeros(len(X_test))

for fold, (train_idx, valid_idx) in enumerate(skf.split(X, y)):

    print("=" * 60)
    print(f"CATBOOST FOLD {fold + 1}")
    print("=" * 60)

    X_train = X.iloc[train_idx]
    X_valid = X.iloc[valid_idx]

    y_train = y.iloc[train_idx]
    y_valid = y.iloc[valid_idx]

    model = CatBoostClassifier(
        iterations=3000,
        learning_rate=0.03,
        depth=7,
        loss_function="Logloss",
        eval_metric="AUC",
        l2_leaf_reg=5,
        random_strength=0.5,
        random_seed=42 + fold,
        verbose=False,
        allow_writing_files=False,
    )

    model.fit(
        X_train,
        y_train,
        cat_features=categorical_cols,
        eval_set=(X_valid, y_valid),
        early_stopping_rounds=200,
        verbose=False,
    )

    oof_cat[valid_idx] = model.predict_proba(X_valid)[:, 1]

    test_cat += model.predict_proba(X_test)[:, 1] / skf.n_splits

    fold_auc = roc_auc_score(y_valid, oof_cat[valid_idx])

    print(f"Fold AUC: {fold_auc:.6f}")

cat_auc = roc_auc_score(y, oof_cat)

print()
print("=" * 60)
print(f"CATBOOST OOF AUC: {cat_auc:.6f}")
print("=" * 60)


# ============================================================
# 9. ONE-HOT ENCODING FOR LIGHTGBM / XGBOOST
# ============================================================

X_encoded = pd.get_dummies(X, columns=categorical_cols, dtype=np.int8)

X_test_encoded = pd.get_dummies(X_test, columns=categorical_cols, dtype=np.int8)

X_test_encoded = X_test_encoded.reindex(columns=X_encoded.columns, fill_value=0)


# ============================================================
# 10. LIGHTGBM
# ============================================================

oof_lgb = np.zeros(len(X_encoded))
test_lgb = np.zeros(len(X_test_encoded))

for fold, (train_idx, valid_idx) in enumerate(skf.split(X_encoded, y)):

    print("=" * 60)
    print(f"LIGHTGBM FOLD {fold + 1}")
    print("=" * 60)

    X_train = X_encoded.iloc[train_idx]
    X_valid = X_encoded.iloc[valid_idx]

    y_train = y.iloc[train_idx]
    y_valid = y.iloc[valid_idx]

    model = LGBMClassifier(
        n_estimators=3000,
        learning_rate=0.02,
        num_leaves=31,
        max_depth=-1,
        min_child_samples=20,
        subsample=0.85,
        colsample_bytree=0.85,
        reg_alpha=0.1,
        reg_lambda=1.0,
        objective="binary",
        random_state=42 + fold,
        n_jobs=-1,
        verbosity=-1,
    )

    model.fit(X_train, y_train, eval_set=[(X_valid, y_valid)], callbacks=[])

    oof_lgb[valid_idx] = model.predict_proba(X_valid)[:, 1]

    test_lgb += model.predict_proba(X_test_encoded)[:, 1] / skf.n_splits

    fold_auc = roc_auc_score(y_valid, oof_lgb[valid_idx])

    print(f"Fold AUC: {fold_auc:.6f}")

lgb_auc = roc_auc_score(y, oof_lgb)

print()
print("=" * 60)
print(f"LIGHTGBM OOF AUC: {lgb_auc:.6f}")
print("=" * 60)


# ============================================================
# 11. XGBOOST
# ============================================================

oof_xgb = np.zeros(len(X_encoded))
test_xgb = np.zeros(len(X_test_encoded))

for fold, (train_idx, valid_idx) in enumerate(skf.split(X_encoded, y)):

    print("=" * 60)
    print(f"XGBOOST FOLD {fold + 1}")
    print("=" * 60)

    X_train = X_encoded.iloc[train_idx]
    X_valid = X_encoded.iloc[valid_idx]

    y_train = y.iloc[train_idx]
    y_valid = y.iloc[valid_idx]

    model = XGBClassifier(
        n_estimators=2500,
        learning_rate=0.02,
        max_depth=5,
        min_child_weight=3,
        subsample=0.85,
        colsample_bytree=0.85,
        reg_alpha=0.1,
        reg_lambda=2.0,
        objective="binary:logistic",
        eval_metric="auc",
        tree_method="hist",
        random_state=42 + fold,
        n_jobs=-1,
    )

    model.fit(X_train, y_train, eval_set=[(X_valid, y_valid)], verbose=False)

    oof_xgb[valid_idx] = model.predict_proba(X_valid)[:, 1]

    test_xgb += model.predict_proba(X_test_encoded)[:, 1] / skf.n_splits

    fold_auc = roc_auc_score(y_valid, oof_xgb[valid_idx])

    print(f"Fold AUC: {fold_auc:.6f}")

xgb_auc = roc_auc_score(y, oof_xgb)

print()
print("=" * 60)
print(f"XGBOOST OOF AUC: {xgb_auc:.6f}")
print("=" * 60)


# ============================================================
# 12. INDIVIDUAL MODEL SCORES
# ============================================================

print("\n")
print("=" * 60)
print("INDIVIDUAL MODEL SCORES")
print("=" * 60)

print(f"CatBoost : {cat_auc:.6f}")
print(f"LightGBM : {lgb_auc:.6f}")
print(f"XGBoost  : {xgb_auc:.6f}")


# ============================================================
# 13. BLENDING
# ============================================================

weights = [
    (0.50, 0.30, 0.20),
    (0.40, 0.40, 0.20),
    (0.40, 0.30, 0.30),
    (0.30, 0.40, 0.30),
    (0.30, 0.30, 0.40),
    (0.50, 0.25, 0.25),
    (0.45, 0.35, 0.20),
    (0.45, 0.30, 0.25),
    (0.35, 0.35, 0.30),
]

best_score = 0
best_weights = None
best_oof = None

print("\n")
print("=" * 60)
print("BLEND SEARCH")
print("=" * 60)

for w_cat, w_lgb, w_xgb in weights:

    blend = w_cat * oof_cat + w_lgb * oof_lgb + w_xgb * oof_xgb

    score = roc_auc_score(y, blend)

    print(
        f"Cat={w_cat:.2f} | "
        f"LGB={w_lgb:.2f} | "
        f"XGB={w_xgb:.2f} | "
        f"AUC={score:.6f}"
    )

    if score > best_score:

        best_score = score

        best_weights = (w_cat, w_lgb, w_xgb)

        best_oof = blend.copy()


# ============================================================
# 14. BEST BLEND
# ============================================================

print("\n")
print("=" * 60)
print("BEST BLEND")
print("=" * 60)

print(f"CatBoost weight : {best_weights[0]:.2f}")

print(f"LightGBM weight : {best_weights[1]:.2f}")

print(f"XGBoost weight  : {best_weights[2]:.2f}")

print(f"OOF ROC-AUC     : {best_score:.6f}")


# ============================================================
# 15. FINAL TEST PREDICTION
# ============================================================

final_test_pred = (
    best_weights[0] * test_cat + best_weights[1] * test_lgb + best_weights[2] * test_xgb
)


# ============================================================
# 16. CREATE SUBMISSION
# ============================================================

submission = pd.DataFrame({"id": test["id"], "satisfaction": final_test_pred})

submission.to_csv("/kaggle/working/submission.csv", index=False)

print("\n")
print("=" * 60)
print("SUBMISSION CREATED")
print("=" * 60)

print(submission.head())
print(submission.shape)

print("\nSaved to: /kaggle/working/submission.csv")
```
