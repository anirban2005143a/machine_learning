```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_excel("ENB2012_data.xlsx")
df.columns = ["Relative Compactness" , "Surface Area" , "Wall Area" , "Roof Area" , "Overall Height" , "Orientation" , "Glazing Area" , "Glazing Area Distribution" , "Heating Load" , "Cooling Load"] 
df.head(5)
```

```python
for cl in df.columns.values :
    plt.figure(figsize=(3,3))
    sns.scatterplot(x=df[cl] , y=df["Heating Load"])
    plt.show()
    plt.figure(figsize=(3,3))
    sns.scatterplot(x=df[cl] , y=df["Cooling Load"])
    plt.show()
```

```python
corr = df.corr(numeric_only=True)

plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```
