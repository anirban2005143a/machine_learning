```python
from string import punctuation
from collections import Counter
```

```python
textfile = open(r"D:\OneDrive - Indian Institute of Technology Indian School of Mines Dhanbad\Downloads\The quiet rhythm of early morning o.txt" , "r" , encoding="utf-8")
textdata = textfile.read()
textdata
```

```python
punctuation
```

```python
table = str.maketrans('', '', punctuation)
word_doc = textdata.translate(table).lower().split()
word_doc
```

```python
word_counts = Counter(word_doc)
for word in word_counts.most_common(15) : 
    print(word)
```
